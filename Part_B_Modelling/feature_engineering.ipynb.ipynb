```python
import pandas as pd

# Load the BankChurners dataset
df = pd.read_csv("BankChurners.csv")

# Display the first five rows
df.head()
```

```python
# Remove irrelevant columns

columns_to_remove = [
    "CLIENTNUM",
    "Naive_Bayes_Classifier_Attrition_Flag_Card_Category_Contacts_Count_12_mon_Dependent_count_Education_Level_Months_Inactive_12_mon_1",
    "Naive_Bayes_Classifier_Attrition_Flag_Card_Category_Contacts_Count_12_mon_Dependent_count_Education_Level_Months_Inactive_12_mon_2"
]

df = df.drop(columns=columns_to_remove, errors="ignore")

print("Remaining columns:", df.shape[1])
```

```python
# Create average transaction amount

df["Avg_Transaction_Amount"] = (
    df["Total_Trans_Amt"] / df["Total_Trans_Ct"]
)

# Display the new feature
print(df[["Total_Trans_Amt", "Total_Trans_Ct", "Avg_Transaction_Amount"]].head())
```

```python
# Check the new feature for missing and infinite values

print("Missing values:")
print(df["Avg_Transaction_Amount"].isnull().sum())

print("\nInfinite values:")
print(df["Avg_Transaction_Amount"].isin([float("inf"), float("-inf")]).sum())
```

```python
# Create high inactivity indicator

df["High_Inactivity"] = (
    df["Months_Inactive_12_mon"] >= 3
).astype(int)

# Check the new feature
print(df["High_Inactivity"].value_counts())
```

```python
# Create transaction activity feature

df["Transaction_Activity"] = (
    df["Total_Trans_Ct"] * df["Total_Trans_Amt"]
)

# Display the new feature
print(
    df[
        ["Total_Trans_Ct", "Total_Trans_Amt", "Transaction_Activity"]
    ].head()
)
```

```python
# Check the final feature set

print("Total number of columns:", df.shape[1])

print("\nEngineered features:")
print([
    "Avg_Transaction_Amount",
    "High_Inactivity",
    "Transaction_Activity"
])

print("\nNew feature preview:")
print(
    df[
        [
            "Avg_Transaction_Amount",
            "High_Inactivity",
            "Transaction_Activity"
        ]
    ].head()
)
```

```python
# Save the feature-engineered dataset

df.to_csv("feature_engineered.csv", index=False)

print("Feature-engineered dataset saved successfully.")
```
