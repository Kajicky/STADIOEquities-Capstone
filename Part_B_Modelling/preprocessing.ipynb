```python
import pandas as pd

# Load the BankChurners dataset
df = pd.read_csv("BankChurners.csv")

# Display the first five rows
df.head()
```

```python
# Check the size of the dataset
print("Dataset shape:", df.shape)

# Check the column names and data types
df.info()
```

```python
# Check for missing values in each column
missing_values = df.isnull().sum()

print(missing_values)
```

```python
# Display all column names
print(df.columns.tolist())
```

```python
# Check the customer churn classes
print(df["Attrition_Flag"].value_counts())
```

```python
# Convert the target variable into binary values
df["Attrition_Flag"] = df["Attrition_Flag"].map({
    "Existing Customer": 0,
    "Attrited Customer": 1
})

# Check the result
print(df["Attrition_Flag"].value_counts())
```

```python
# Separate features (X) and target (y)

X = df.drop(columns=["Attrition_Flag"])
y = df["Attrition_Flag"]

print("Features:", X.shape)
print("Target:", y.shape)
```

```python
# Identify categorical columns

categorical_columns = X.select_dtypes(include=["object"]).columns.tolist()

print("Categorical columns:")
print(categorical_columns)
```

```python
# Convert categorical variables into numerical variables

X = pd.get_dummies(X, columns=categorical_columns, drop_first=True)

# Check the new dataset
print("Number of features after encoding:", X.shape[1])
print(X.head())
```

```python
# Check the data types of all features
print(X.dtypes.value_counts())
```

```python
from sklearn.model_selection import train_test_split

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
# Check the class distribution in the training data
print("Training class distribution:")
print(y_train.value_counts())

print("\nTraining class percentages:")
print(y_train.value_counts(normalize=True) * 100)
```

```python
from sklearn.preprocessing import StandardScaler

# Create the scaler
scaler = StandardScaler()

# Fit the scaler only on the training data
X_train_scaled = scaler.fit_transform(X_train)

# Apply the same scaling to the test data
X_test_scaled = scaler.transform(X_test)

print("Training data scaled:", X_train_scaled.shape)
print("Testing data scaled:", X_test_scaled.shape)
```

```python
# Save the processed datasets

train_data = pd.DataFrame(X_train_scaled, columns=X.columns)
train_data["Attrition_Flag"] = y_train.reset_index(drop=True)

test_data = pd.DataFrame(X_test_scaled, columns=X.columns)
test_data["Attrition_Flag"] = y_test.reset_index(drop=True)

train_data.to_csv("train_processed.csv", index=False)
test_data.to_csv("test_processed.csv", index=False)

print("Processed training and testing datasets saved successfully.")
```
